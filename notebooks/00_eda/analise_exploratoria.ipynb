## Carregar e Visualizar o Conjunto de Dados

Carregue o dataset desejado (ajuste o caminho conforme necessário).

```python
# Substitua pelo caminho do seu arquivo de dados
caminho_arquivo = '../data/final/seu_arquivo.csv'

df = pd.read_csv(caminho_arquivo)
df.head()
```

## Análise de Dados Categóricos

Vamos analisar as variáveis categóricas do dataset.

```python
# Selecionar colunas categóricas
dados_categoricos = df.select_dtypes(include=['object', 'category'])

for coluna in dados_categoricos.columns:
    print(f'Coluna: {coluna}')
    print(dados_categoricos[coluna].value_counts())
    print('\n')
    plt.figure(figsize=(8,4))
    sns.countplot(data=df, x=coluna, order=dados_categoricos[coluna].value_counts().index)
    plt.title(f'Distribuição de {coluna}')
    plt.xticks(rotation=45)
    plt.show()
```

## Análise de Dados Numéricos

Vamos analisar as variáveis numéricas do dataset.

```python
# Selecionar colunas numéricas
dados_numericos = df.select_dtypes(include=[np.number])

# Estatísticas descritivas
dados_numericos.describe().T
```

```python
# Histogramas e boxplots para variáveis numéricas
for coluna in dados_numericos.columns:
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    sns.histplot(df[coluna], kde=True, ax=axes[0])
    axes[0].set_title(f'Histograma de {coluna}')
    sns.boxplot(x=df[coluna], ax=axes[1])
    axes[1].set_title(f'Boxplot de {coluna}')
    plt.show()
```

## Análise de Correlação

Vamos calcular e visualizar a matriz de correlação entre as variáveis numéricas.

```python
# Matriz de correlação
matriz_corr = dados_numericos.corr()

plt.figure(figsize=(10,8))
sns.heatmap(matriz_corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Matriz de Correlação')
plt.show()
```

## Análise de Variáveis Dependentes com o Target

Explore como as variáveis independentes se relacionam com a variável target.

```python
# Substitua 'target' pelo nome da sua variável alvo
target = 'target'

# Variáveis categóricas x target
for coluna in dados_categoricos.columns:
    if coluna != target:
        plt.figure(figsize=(8,4))
        sns.boxplot(x=coluna, y=target, data=df)
        plt.title(f'{coluna} vs {target}')
        plt.xticks(rotation=45)
        plt.show()

# Variáveis numéricas x target
for coluna in dados_numericos.columns:
    if coluna != target:
        plt.figure(figsize=(6,4))
        sns.scatterplot(x=coluna, y=target, data=df)
        plt.title(f'{coluna} vs {target}')
        plt.show()
```
