# Análise Exploratória de Dados (EDA) - Brazilian E-commerce

Este notebook realiza uma exploração automatizada de todos os datasets disponíveis na pasta `data/raw`.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
from pathlib import Path

# Configurações de visualização
plt.style.use('seaborn-v0_8')
sns.set_palette('pastel')
%matplotlib inline

RAW_DATA_PATH = Path('../../data/raw')
files = [f for f in os.listdir(RAW_DATA_PATH) if f.endswith('.csv')]
print(f'Arquivos encontrados: {len(files)}')
```

## Resumo Geral dos Datasets
Vamos percorrer cada arquivo e exibir as primeiras linhas, informações de tipos e valores nulos.

```python
datasets = {}

for file in files:
    df_name = file.replace('.csv', '')
    df = pd.read_csv(RAW_DATA_PATH / file)
    datasets[df_name] = df
    
    print(f'\n' + '='*50)
    print(f'DATASET: {df_name}')
    print('='*50)
    print(f'Shape: {df.shape}')
    display(df.head(2))
    
    print('\nInformações do Dataset:')
    display(df.info())
    
    print('\nValores Nulos:')
    display(df.isnull().sum()[df.isnull().sum() > 0])
    
    print('\nEstatísticas Descritivas (Numéricas):')
    display(df.describe().T)
```

## Visualização de Dados Críticos
Nesta seção, focamos em distribuições de preços e pagamentos.

```python
if 'olist_order_items_dataset' in datasets:
    df_items = datasets['olist_order_items_dataset']
    plt.figure(figsize=(10, 5))
    sns.histplot(df_items['price'], bins=50, kde=True)
    plt.title('Distribuição de Preços dos Itens')
    plt.show()

if 'olist_order_payments_dataset' in datasets:
    df_payments = datasets['olist_order_payments_dataset']
    plt.figure(figsize=(8, 5))
    sns.countplot(data=df_payments, x='payment_type')
    plt.title('Tipos de Pagamento mais Comuns')
    plt.show()
```
